```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
circuito_3q = QuantumCircuit(3, 3)
circuito_3q.h(0)
circuito_3q.cx(0, 1)
circuito_3q.cx(1, 2)
circuito_3q.measure([0, 1, 2], [0, 1, 2])
display(circuito_3q.draw(output='mpl'))
simulador = AerSimulator()
resultado = simulador.run(circuito_3q).result()
cuentas = resultado.get_counts()
plot_histogram(cuentas)
```
